# TWTT time-series analysis

This notebook plots offset, combined time of flight, per-way time of flight, and per-way coincidence rates. Per-way time of flight is `abs(corrected_peak_ps)`, where `corrected_peak_ps = peak_position_ps + configured_peak0_ps`. The elapsed-time x-axis automatically uses seconds, minutes, or hours based on the acquisition duration.

```python
import sys
from pathlib import Path
from pprint import pprint

# Support launching Jupyter from either the repository root or notebooks/.
repository_root = next(
    path for path in (Path.cwd(), Path.cwd().parent)
    if (path / 'src' / 'twtt_data').is_dir()
)
source_directory = repository_root / 'src'
if str(source_directory) not in sys.path:
    sys.path.insert(0, str(source_directory))

from twtt_data.analysis import (
    load_twtt_jsonl,
    plot_twtt,
    plot_twtt_smoothed,
    summarize,
)

%matplotlib inline
```

```python
data_directory = repository_root / 'data'
if not data_directory.is_dir():
    raise FileNotFoundError(f'Data directory not found: {data_directory}')
data_files = sorted(data_directory.glob('twtt_*.jsonl'))
if not data_files:
    raise FileNotFoundError('No twtt_*.jsonl file found')

data_file = data_files[-1]
metadata, samples = load_twtt_jsonl(data_file)
print(f'Loaded {len(samples):,} results from {data_file.name}')
print(f"From {samples[0]['timestamp']} to {samples[-1]['timestamp']}")
```

```python
pprint(summarize(samples), sort_dicts=False)
```

```python
figure, axes = plot_twtt(samples)
figure.suptitle(f'TWTT acquisition — {data_file.name}', fontsize=14)
```

```python
smooth_figure, smooth_axes = plot_twtt_smoothed(samples, window_size=100)
smooth_figure.suptitle(
    f'TWTT acquisition — 100-sample moving average — {data_file.name}',
    fontsize=14,
)
```
